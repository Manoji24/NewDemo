```python
%sql
DROP TABLE IF EXISTS capstone.silver.products;
CREATE TABLE capstone.silver.products
USING DELTA
AS
SELECT
  item_id,
  product_name,
  initcap(lower(category)) AS category,
  _ingest_timestamp,
  _source_file_name
FROM capstone.bronze.products;
```
