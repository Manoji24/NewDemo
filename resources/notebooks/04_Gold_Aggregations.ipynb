```python
%sql
CREATE OR REPLACE TABLE capstone.gold.daily_sales_summary
USING DELTA
AS
SELECT
  order_date,
  region,
  SUM(line_total) AS total_revenue,
  SUM(quantity) AS total_units_sold,
  ROUND(SUM(line_total) / NULLIF(COUNT(DISTINCT order_id), 0), 2) AS avg_order_value,
  COUNT(DISTINCT order_id) AS orders_count
FROM capstone.gold.daily_sales_fact
GROUP BY order_date, region;
```

```python
%sql
SELECT * FROM capstone.gold.daily_sales_summary;
```
